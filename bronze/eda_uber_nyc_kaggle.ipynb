```python
#Importing Files from Kaggle
import kagglehub

# Download latest version
path = kagglehub.dataset_download("fivethirtyeight/uber-pickups-in-new-york-city")

print("Path to dataset files:", path)
```

```python
import os

print("Files downloaded:")
for f in os.listdir(path):
    file_path = os.path.join(path, f)
    size_mb = os.path.getsize(file_path) / (1024 * 1024)
    print(f"{f} - {size_mb:.2f} MB")
```

```python
import pandas as pd

sample_2014 = pd.read_csv(os.path.join(path, "uber-raw-data-apr14.csv"), nrows=5)
print("2014 file columns:")
print(sample_2014.columns.tolist())
print(sample_2014.head())

sample_2015 = pd.read_csv(os.path.join(path, "uber-raw-data-janjune-15.csv"), nrows=5)
print("\n2015 file columns:")
print(sample_2015.columns.tolist())
print(sample_2015.head())
```
